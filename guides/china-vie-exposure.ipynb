# China exposure: VIE structures and HFCAA risk in filings

Find filers that operate in China through a variable interest entity, and those discussing delisting risk under the HFCAA.

Read it with charts at https://www.quantisnow.com/sec-api/guides/china-vie-exposure. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Many China-based companies listed in the US use a variable interest entity: holders own a Cayman or BVI company that controls the Chinese business through contracts, not shares. The Holding Foreign Companies Accountable Act adds a delisting risk if the auditor can't be inspected. Both are disclosed in 10-Ks, 20-Fs and IPO prospectuses.

## Filers with a China VIE structure

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"variable interest entity\" AND (PRC OR \"People's Republic of China\")",
        "formTypes": ["10-K","10-Q","20-F","F-1","F-1/A"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````
