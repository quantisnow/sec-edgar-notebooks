# Executive pay: search proxy statements (DEF 14A)

Find proxy statements with the summary compensation table, pay versus performance, and say-on-pay results.

Read it with charts at https://www.quantisnow.com/sec-api/guides/executive-pay-def-14a. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

The proxy statement for the annual meeting (DEF 14A) has the summary compensation table: salary, bonus, stock and option awards, and total pay for the CEO, CFO and the next three. Since 2023 it also has pay versus performance, which compares compensation actually paid with shareholder return.

## Proxy statements with a summary compensation table

### Which proxy statements have a summary compensation table?

Every DEF 14A for an annual meeting has the table: salary, bonus, stock awards and total pay for the top executives.

```shell
curl -X POST https://www.quantisnow.com/api/v1/full-text \
  -H "Authorization: Bearer $QUANTISNOW_API_KEY" \
  -H "Content-Type: application/json" \
  -d '{
  "query": "\"summary compensation table\"",
  "formTypes": ["DEF 14A"],
  "sort": "filedAt:desc",
  "size": 5
}'
```

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"summary compensation table\"",
        "formTypes": ["DEF 14A"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## From the command line

```shell
npx quantisnow search '"summary compensation table"' --forms 'DEF 14A' --newest
```

Say-on-pay results come later, in an 8-K under Item 5.07 after the meeting: `npx quantisnow events votes`.
