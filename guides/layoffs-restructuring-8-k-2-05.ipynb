# Layoffs and restructuring: 8-K Item 2.05

Companies that committed to job cuts, closures or exit plans with material charges, and the cost estimates they gave.

Read it with charts at https://www.quantisnow.com/sec-api/guides/layoffs-restructuring-8-k-2-05. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Item 2.05 is required when a board commits to an exit or disposal plan with material charges: severance, lease exits, asset write-offs. The 8-K gives the expected charge by type and how much of it is cash.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/8-k/events",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "items.item:\"2.05\" AND filedAt:[2026-01-01 TO 2026-11-24]",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
page = res.json()
print(page["total"]["value"], "matches")
for f in page["data"]:
    print(f["filedAt"][:10], f["ticker"], f["companyName"])
````
