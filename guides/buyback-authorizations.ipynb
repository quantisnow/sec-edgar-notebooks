# Share buyback authorizations in 8-K filings

Find boards that authorized a new or larger share repurchase program, with the dollar amount in the passage.

Read it with charts at https://www.quantisnow.com/sec-api/guides/buyback-authorizations. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

A repurchase authorization is permission, not a purchase: some companies spend all of it, some none. Size it against market value and check later 10-Qs, which show shares actually bought each month.

## New buyback authorizations in 8-Ks

### Which boards authorized a new share repurchase program?

A new or larger buyback authorization says the board has cash it doesn't need elsewhere. The size relative to market value matters more than the headline.

```shell
curl -X POST https://www.quantisnow.com/api/v1/full-text \
  -H "Authorization: Bearer $QUANTISNOW_API_KEY" \
  -H "Content-Type: application/json" \
  -d '{
  "query": "\"share repurchase program\"~3 AND (authorized OR approved)",
  "formTypes": ["8-K"],
  "sort": "filedAt:desc",
  "size": 5
}'
```

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"share repurchase program\"~3 AND (authorized OR approved)",
        "formTypes": ["8-K"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````
