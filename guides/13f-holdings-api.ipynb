# 13F holdings API: any fund's portfolio and what changed

Pull a manager's 13F positions by quarter, or every fund holding a stock, with the change from the quarter before already worked out. Python and JavaScript.

Read it with charts at https://www.quantisnow.com/sec-api/guides/13f-holdings-api. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Managers with more than $100 million in US equities file a 13F-HR within 45 days of each quarter end. The filing lists positions by CUSIP, with no tickers and no comparison to the last quarter. The [13F endpoint](https://www.quantisnow.com/sec-api/13f-holdings) maps CUSIPs to tickers and keeps last quarter's shares on every record, so `change:new` or `change:sold` is a filter.

## One fund's portfolio

Search by `manager.cik`, which stays the same when a fund changes its name, and sort by value.

### What were Berkshire Hathaway's largest holdings last quarter?

A whole portfolio in one call, ranked by size.

```shell
curl -X POST https://www.quantisnow.com/api/v1/13f/holdings \
  -H "Authorization: Bearer $QUANTISNOW_API_KEY" \
  -H "Content-Type: application/json" \
  -d '{
  "query": "manager.cik:0001067983 AND period:2026-06-30",
  "from": 0,
  "size": 5,
  "sort": [{"valueUsd": {"order": "desc"}}]
}'
```

## A portfolio into pandas

````python
import os, requests

BASE = "https://www.quantisnow.com/api/v1"
KEY = os.environ["QUANTISNOW_API_KEY"]

def search(endpoint, query, size=50, sort=None, start=0):
    body = {"query": query, "size": size, "from": start}
    if sort:
        body["sort"] = sort
    res = requests.post(f"{BASE}/{endpoint}", json=body,
                        headers={"Authorization": f"Bearer {KEY}"}, timeout=30)
    res.raise_for_status()
    return res.json()
import pandas as pd

rows, start = [], 0
while True:
    page = search("13f/holdings", "manager.cik:0001067983 AND period:2026-06-30",
                  sort="valueUsd:desc", start=start)
    rows += page["data"]
    start += 50
    if start >= page["total"]["value"]:
        break

df = pd.json_normalize(rows)[["issuer.ticker", "issuer.name", "shares", "valueUsd", "change", "changeValueUsd"]]
df["weight"] = df["valueUsd"] / df["valueUsd"].sum()
print(df.head(10))
````

## In JavaScript

```javascript
const BASE = "https://www.quantisnow.com/api/v1";

async function search(endpoint, query, { size = 50, sort, from = 0 } = {}) {
  const res = await fetch(BASE + "/" + endpoint, {
    method: "POST",
    headers: {
      Authorization: "Bearer " + process.env.QUANTISNOW_API_KEY,
      "Content-Type": "application/json",
    },
    body: JSON.stringify({ query, size, from, sort }),
  });
  if (!res.ok) throw new Error((await res.json()).error.message);
  return res.json();
}

// Funds that added to an existing Apple position, biggest dollar increase first
const page = await search(
  "13f/holdings",
  "issuer.ticker:AAPL AND change:added AND period:2026-06-30",
  { sort: "changeValueUsd:desc", size: 10 },
);
for (const r of page.data) console.log(r.manager.name, r.changeValueUsd);
```

`period` is the quarter end the holdings are for, not the filing date. Filings for a quarter keep arriving for weeks after it ends and amendments come later still, so a quarter is complete a few days after the 45-day deadline.
