# New 13F positions from well-known managers

Pull a well-known manager's 13F by CIK and keep only the new positions and the biggest adds, with the change from last quarter worked out.

Read it with charts at https://www.quantisnow.com/sec-api/guides/famous-investors-13f-new-positions. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Filtering a manager's 13F on `change:new` gives the stocks it didn't own a quarter earlier. Use the manager's CIK, not its name: names change and get abbreviated, CIKs don't. Berkshire Hathaway is 0001067983.

## Berkshire Hathaway's 13F, largest first

### What were Berkshire Hathaway's largest holdings last quarter?

A whole portfolio in one call, ranked by size.

```shell
curl -X POST https://www.quantisnow.com/api/v1/13f/holdings \
  -H "Authorization: Bearer $QUANTISNOW_API_KEY" \
  -H "Content-Type: application/json" \
  -d '{
  "query": "manager.cik:0001067983 AND period:2026-06-30",
  "from": 0,
  "size": 5,
  "sort": [{"valueUsd": {"order": "desc"}}]
}'
```

## In Python

````python
import os, requests

MANAGERS = {"Berkshire Hathaway": "0001067983"}  # add the CIKs you follow
H = {"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"}

for name, cik in MANAGERS.items():
    res = requests.post("https://www.quantisnow.com/api/v1/13f/holdings", headers=H, json={
        "query": f"manager.cik:{cik} AND period:2026-06-30 AND change:(new OR added)",
        "sort": "changeValueUsd:desc", "size": 20,
    }).json()
    for h in res["data"]:
        print(name, h["change"], h["issuer"]["ticker"], round(h["changeValueUsd"]))
````

## From the command line

```shell
npx quantisnow query 13f/holdings 'manager.cik:0001067983 AND period:2026-06-30' --sort valueUsd:desc
```

A 13F shows holdings 45 days after the quarter ended, so a new position may already be bigger, smaller or gone. Look up a manager's CIK on its 13F filing on sec.gov, or search `manager.name:Name*`.
