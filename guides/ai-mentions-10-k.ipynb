# AI mentions in 10-K filings

Find annual reports that talk about artificial intelligence, as a product, a cost or a risk, with counts by form and company.

Read it with charts at https://www.quantisnow.com/sec-api/guides/ai-mentions-10-k. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Counting AI mentions says little; reading where they sit says a lot. In a product section it's a bet; in risk factors it's a worry, often about competitors or regulation; in MD&A it's spending.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"artificial intelligence\" OR \"generative AI\"",
        "formTypes": ["10-K","10-K/A"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## From the command line

```shell
npx quantisnow search '"artificial intelligence" OR "generative AI"' --forms 10-K --newest
```

Add `facets: true` to the API body to count matches by company, or narrow to risk language with `"artificial intelligence" AND "could adversely affect"~10`.
