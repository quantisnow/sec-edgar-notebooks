# Delisting notices: 8-K Item 3.01 from Nasdaq and NYSE

Companies that got a notice from their exchange for a low share price, late filings or too little equity, with Python and CLI to track them.

Read it with charts at https://www.quantisnow.com/sec-api/guides/delisting-notices-8-k-3-01. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Item 3.01 covers a notice that a company no longer meets a listing rule, and the transfer to another market. The most common is Nasdaq's $1 minimum bid price: thirty business days under $1 starts a 180-day clock, and a reverse split is the usual way out.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/8-k/events",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "items.item:\"3.01\" AND filedAt:[2026-10-25 TO 2026-11-24]",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
page = res.json()
print(page["total"]["value"], "matches")
for f in page["data"]:
    print(f["filedAt"][:10], f["ticker"], f["companyName"])
````
