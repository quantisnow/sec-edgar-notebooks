# Activist 13D filings that say what the holder plans to do

Search Schedule 13D text for letters to the board, director nominations and proxy contests, not just the 5% threshold.

Item 4 of Schedule 13D, Purpose of Transaction, is where a 5% holder says what it intends. Most use stock wording that reserves every right without committing to any. The filings worth reading are the ones that say something has happened: a letter delivered, a nomination notice, a proxy contest.

## 13Ds that report a letter, a nomination or a proxy fight

### Which 13D filings say the holder sent a letter, plans to nominate directors or is in a proxy contest?

Item 4 of Schedule 13D, Purpose of Transaction, is where a holder says what it plans to do. Most 13Ds reserve the right to do everything in stock language; these phrases find the holders who say they've started.

```shell
curl -X POST https://www.quantisnow.com/api/v1/full-text \
  -H "Authorization: Bearer $QUANTISNOW_API_KEY" \

  -d '{
  "query": "\"intend to nominate\"~3 OR \"intends to nominate\"~3 OR \"delivered a letter\"~4 OR \"nomination notice\"~3 OR \"proxy contest\"",
  "formTypes": ["SCHEDULE 13D", "SCHEDULE 13D/A"],
  "sort": "filedAt:desc",
  "size": 5
}'
```

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"intend to nominate\"~3 OR \"intends to nominate\"~3 OR \"delivered a letter\"~4 OR \"nomination notice\"~3 OR \"proxy contest\"",
        "formTypes": ["SCHEDULE 13D","SCHEDULE 13D/A"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## From the command line

```shell
npx quantisnow search '"delivered a letter" OR "intends to nominate" OR "proxy contest"' --forms 'SCHEDULE 13D,SCHEDULE 13D/A' --newest
```

The letter itself is usually an exhibit to the 13D, and full-text search reads exhibits. Add the holder's name to find every letter it has sent.
