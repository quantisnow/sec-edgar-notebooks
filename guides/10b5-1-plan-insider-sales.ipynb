# 10b5-1 trading plans: separate planned sales from the rest

Split insider sales into those made under a Rule 10b5-1 plan and those decided on the spot, with the Form 4 flag the API exposes.

Read it with charts at https://www.quantisnow.com/sec-api/guides/10b5-1-plan-insider-sales. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

A 10b5-1 plan is set up in advance, with a cooling-off period, and then sells on a schedule. Since 2023 Form 4 has a checkbox for it, and every 10-Q and 10-K has to say which officers and directors adopted or ended a plan. The API carries the Form 4 flag as `rule10b51`.

## Large sales made under a 10b5-1 plan

### Which insider sales over $5 million in the last 30 days were made under a 10b5-1 plan?

Sales under a plan set up months earlier say little about today's view. Separate them before reading anything into a sale.

```shell
curl -X POST https://www.quantisnow.com/api/v1/insider-trades \
  -H "Authorization: Bearer $QUANTISNOW_API_KEY" \
  -H "Content-Type: application/json" \
  -d '{
  "query": "formType:4 AND rule10b51:true AND sold.valueUsd:>5000000 AND filedAt:[2026-10-25 TO 2026-11-24]",
  "from": 0,
  "size": 5,
  "sort": [{"sold.valueUsd": {"order": "desc"}}]
}'
```

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/insider-trades",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "formType:4 AND transactionCodes:S AND rule10b51:false AND sold.valueUsd:>1000000 AND filedAt:[2026-10-25 TO 2026-11-24]",
        "sort": "sold.valueUsd:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
page = res.json()
print(page["total"]["value"], "matches")
for r in page["data"]:
    print(r["issuer"]["ticker"], r["reportingOwner"]["name"], round(r["sold"]["valueUsd"]))
````

## From the command line

```shell
npx quantisnow query insider-trades 'formType:4 AND transactionCodes:S AND rule10b51:false AND sold.valueUsd:>1000000' --sort sold.valueUsd:desc
```
