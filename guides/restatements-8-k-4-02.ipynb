# Restatements: find 8-K Item 4.02 non-reliance filings

Item 4.02 says earlier financial statements should no longer be relied upon. Find them by date and company, and read which periods and why.

Read it with charts at https://www.quantisnow.com/sec-api/guides/restatements-8-k-4-02. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

When a company or its auditor finds an error big enough that past numbers can't be trusted, the company files Item 4.02 within four business days. It names the periods, says whether the auditor agreed, and usually describes the error. A restatement follows, sometimes months later.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "items:\"4.02\" OR \"should no longer be relied upon\"",
        "formTypes": ["8-K","8-K/A"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````
