# Bankruptcy filings: 8-K Item 1.03 and Chapter 11 mentions

Find companies that filed for Chapter 11 or Chapter 7, and the 8-Ks around a sale process or stalking-horse bid.

Read it with charts at https://www.quantisnow.com/sec-api/guides/bankruptcy-filings-8-k-1-03. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Item 1.03 is filed when a company or a major subsidiary enters bankruptcy or receivership. Searching the text for "Chapter 11" also finds the filings around it: debtor-in-possession financing, the stalking-horse bid, the plan and the exit.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "items:\"1.03\" OR \"Chapter 11\"",
        "formTypes": ["8-K"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

