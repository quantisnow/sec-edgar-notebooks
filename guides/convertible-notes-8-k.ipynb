# Convertible note financings in 8-K filings

Find 8-Ks announcing convertible notes, from investment-grade issues to small-cap promissory notes, with Python and CLI.

Read it with charts at https://www.quantisnow.com/sec-api/guides/convertible-notes-8-k. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

A convertible note is debt the holder can turn into stock. A large company sells convertible senior notes at a low coupon and hedges the dilution. A small one sells convertible promissory notes to a few lenders, often with a conversion price that moves with the stock. Both start with an 8-K.

## Convertible note 8-Ks

### Which 8-Ks announce convertible notes?

Convertible notes are debt that can turn into stock. Large companies issue them cheaply; small ones often on terms that dilute holders heavily.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"convertible promissory note\" OR \"convertible senior notes due\"~3 OR \"convertible notes due\"~3",
        "formTypes": ["8-K"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````
