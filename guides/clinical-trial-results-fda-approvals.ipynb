# Clinical trial readouts and FDA approvals from filings

Find 8-Ks reporting whether a trial met its primary endpoint, and FDA approval records, in one place.

Read it with charts at https://www.quantisnow.com/sec-api/guides/clinical-trial-results-fda-approvals. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

For a biotech, a trial readout or an FDA decision is the event. Companies file the announcement on an 8-K, with the release as an exhibit; full-text search finds the wording that matters, met or missed the primary endpoint. FDA approval records themselves are searchable with `kinds: ["fda"]`.

## Trial readouts that met or missed the primary endpoint

### Which 8-Ks report whether a clinical trial met its primary endpoint?

### Which companies announced an FDA approval in an 8-K or 6-K?

Drug and device makers file the approval press release as an 8-K exhibit, foreign ones as a 6-K. A bare "FDA approval" mostly finds forward-looking disclaimers, so this asks for the words companies use when it happened.

```shell
curl -X POST https://www.quantisnow.com/api/v1/full-text \
  -H "Authorization: Bearer $QUANTISNOW_API_KEY" \
  -H "Content-Type: application/json" \
  -d '{
  "query": "\"announces FDA approval\" OR \"announced FDA approval\" OR \"received FDA approval\" OR \"FDA has approved\"",
  "formTypes": ["8-K", "6-K"],
  "size": 5
}'
```

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"positive topline\"~3 OR \"met its primary endpoint\"~3 OR \"did not meet the primary endpoint\"~3 OR \"did not meet its primary endpoint\"~3",
        "formTypes": ["8-K"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````
