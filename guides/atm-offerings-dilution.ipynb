# At-the-market offerings: track dilution from 424B5 filings

Find companies selling stock through ATM programs and other shelf takedowns, from the 424B5 prospectus supplement that starts each one.

Read it with charts at https://www.quantisnow.com/sec-api/guides/atm-offerings-dilution. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

An at-the-market program lets a company sell new shares straight into the market through a broker, a few at a time, at whatever the price is. It's cheap and quiet for the company and steady selling pressure for holders. The program starts with a 424B5 prospectus supplement off an effective shelf.

## ATM programs started by a 424B5

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"at-the-market offering\" OR \"at the market offering\"",
        "formTypes": ["424B5","8-K"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## From the command line

```shell
npx quantisnow search '"at-the-market offering"' --forms 424B5 --newest
npx quantisnow query filings 'formType:424B5 AND ticker:MVIS' --sort filedAt:desc
```

How much was actually sold shows up later, in the next 10-Q's equity note or subsequent events. Search the company's 10-Q for `"at-the-market" AND "net proceeds"` to find it.
