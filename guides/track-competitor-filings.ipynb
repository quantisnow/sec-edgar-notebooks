# Track a competitor's SEC filings

Follow every filing from a set of companies, 8-Ks, 10-Qs, Form 4s and 13Ds about them, and search their text for products and customers.

Read it with charts at https://www.quantisnow.com/sec-api/guides/track-competitor-filings. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Competitive intelligence from filings is mostly reading: a competitor's 10-Q segment commentary, its 8-Ks, who's buying or selling its stock. Start with every filing by ticker, then narrow with full-text search.

## Every filing from one company

### What has Nvidia filed with the SEC in the last 90 days?

A company's whole filing history for a quarter, newest first, to diff against what you already have.

```shell
curl -X POST https://www.quantisnow.com/api/v1/filings \
  -H "Authorization: Bearer $QUANTISNOW_API_KEY" \
  -H "Content-Type: application/json" \
  -d '{
  "query": "ticker:NVDA AND filedAt:[2026-08-26 TO 2026-11-24]",
  "from": 0,
  "size": 5,
  "sort": [{"filedAt": {"order": "desc"}}]
}'
```

## In Python

````python
import os, requests

COMPETITORS = ["AMD", "INTC", "AVGO"]
res = requests.post("https://www.quantisnow.com/api/v1/filings",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={"query": f"ticker:({' OR '.join(COMPETITORS)}) AND formType:(\"8-K\" OR \"10-Q\" OR \"10-K\") AND filedAt:[2026-10-01 TO 2026-11-23]",
          "sort": "filedAt:desc", "size": 50}).json()
for f in res["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["description"])
````
