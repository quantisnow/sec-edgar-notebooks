# Shelf registrations: new S-3 filings

Find companies that registered securities to sell later, the step before an ATM program or a quick offering.

Read it with charts at https://www.quantisnow.com/sec-api/guides/shelf-registrations-s-3. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

A Form S-3 registers securities without selling them. Once the SEC declares it effective, the company can sell off it for three years with only a prospectus supplement. Large companies get automatic shelves (S-3ASR); for small ones, a new shelf is often the sign a raise is coming.

## New S-3 shelf registrations

### Which companies filed a shelf registration to sell securities from time to time?

A shelf (Form S-3) registers securities to sell later, in pieces. It doesn't sell anything by itself, but it's the step before an ATM or a quick offering.

```shell
curl -X POST https://www.quantisnow.com/api/v1/full-text \
  -H "Authorization: Bearer $QUANTISNOW_API_KEY" \
  -H "Content-Type: application/json" \
  -d '{
  "query": "\"from time to time\" AND (shelf OR \"one or more offerings\")",
  "formTypes": ["S-3", "S-3/A", "S-3ASR"],
  "sort": "filedAt:desc",
  "size": 5
}'
```

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"from time to time\" AND (shelf OR \"one or more offerings\")",
        "formTypes": ["S-3","S-3/A","S-3ASR"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## From the command line

```shell
npx quantisnow query filings 'formType:("S-3" OR "S-3ASR") AND filedAt:[2026-11-01 TO 2026-11-23]' --sort filedAt:desc -n 50
```

Read who is selling. A resale shelf registers shares for existing holders, often PIPE investors, which means selling by them rather than new money for the company.
