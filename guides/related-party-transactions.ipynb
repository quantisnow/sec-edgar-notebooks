# Related-party transactions in 10-K and 10-Q filings

Find loans, leases and sales between a company and its insiders, from the related-party notes in quarterly and annual reports.

Read it with charts at https://www.quantisnow.com/sec-api/guides/related-party-transactions. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Item 404 makes companies disclose transactions over $120,000 with directors, officers, large holders and their families. Small companies have the most: the founder's building, a loan from the CEO, sales to a director's other company.

## Related-party loans, leases and balances

### Which filings disclose transactions with related parties?

Loans to executives, leases from the founder, sales to a director's company: Item 404 makes companies list them, and small companies have the most.

```shell
curl -X POST https://www.quantisnow.com/api/v1/full-text \
  -H "Authorization: Bearer $QUANTISNOW_API_KEY" \
  -H "Content-Type: application/json" \
  -d '{
  "query": "\"related party transactions\" AND (loan OR lease OR \"due to\")",
  "formTypes": ["10-Q", "10-Q/A", "10-K", "10-K/A"],
  "sort": "filedAt:desc",
  "size": 5
}'
```

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"related party transactions\" AND (loan OR lease OR \"due to\")",
        "formTypes": ["10-Q","10-Q/A","10-K","10-K/A"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````
