# Customer concentration: companies dependent on one or two customers

Search 10-Ks and 10-Qs for customers above 10% of revenue, with the percentages in the passage.

Read it with charts at https://www.quantisnow.com/sec-api/guides/customer-concentration. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Companies must name, or at least count, customers above 10% of revenue. A supplier whose largest customer is half its sales lives on that contract, and losing it can halve the company.

## Customers above 10% of revenue

### Which companies get a large share of revenue from one or two customers?

Companies have to disclose customers above 10% of revenue. Losing one of them can halve a small company's sales.

```shell
curl -X POST https://www.quantisnow.com/api/v1/full-text \
  -H "Authorization: Bearer $QUANTISNOW_API_KEY" \

  "query": "\"one customer accounted for\"~3 OR \"two customers accounted for\"~3 OR \"largest customer accounted for\"~3",
  "formTypes": ["10-Q", "10-Q/A", "10-K", "10-K/A"],
  "sort": "filedAt:desc",
  "size": 5
}'
```

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"one customer accounted for\"~3 OR \"two customers accounted for\"~3 OR \"largest customer accounted for\"~3",
        "formTypes": ["10-Q","10-Q/A","10-K","10-K/A"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## From the command line

```shell
npx quantisnow search '"one customer accounted for"~3 OR "largest customer accounted for"~3' --forms 10-K,10-Q --newest
```

Pull the percentage out of each passage with a regular expression like `(\d+(?:\.\d+)?)\s?%`, then sort by it.
