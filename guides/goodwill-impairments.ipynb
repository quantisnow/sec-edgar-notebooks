# Goodwill impairment charges in SEC filings

Find companies that wrote down goodwill from past acquisitions, in 10-Qs, 10-Ks and Item 2.06 8-Ks, with the amount in the passage.

Read it with charts at https://www.quantisnow.com/sec-api/guides/goodwill-impairments. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Goodwill is what a company paid for an acquisition above the value of its assets. Writing it down admits the deal is worth less than it paid. It's non-cash, but it often comes with weaker guidance for the business that was bought.

## Item 2.06 impairments

Item 2.06 is required when a company concludes an asset is materially impaired outside a regular report.

  "query": "items.item:\"2.06\" AND filedAt:[2026-01-01 TO 2026-11-24]",

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"recorded a goodwill impairment\"~4 OR \"recognized a goodwill impairment\"~4 OR \"non-cash goodwill impairment charge\"~3",
        "formTypes": ["8-K","10-Q","10-K"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````
