# 8-K Item 5.02 API: CEO, CFO and director departures

Find every 8-K that reports an officer or director leaving or joining, then narrow to CEO and CFO exits with full-text search. Python, JavaScript and real results.

Read it with charts at https://www.quantisnow.com/sec-api/guides/8-k-item-5-02-officer-changes. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Item 5.02 of Form 8-K covers departures and appointments of directors and certain officers, and compensation arrangements with them. A company has four business days to file it. That makes it the fastest public record of a CEO or CFO leaving, but the item number alone also catches every new director and every bonus plan.

## Only the CEO exits

[Full-text search](https://www.quantisnow.com/sec-api/full-text-search) can look inside the filing. Proximity phrases like `"resign chief executive officer"~3` match "resigned as Chief Executive Officer" and "will resign from his role as chief executive officer" without matching boilerplate about a future successor.

### Which CEOs resigned, retired or said they would step down?

Item 5.02 covers every officer and director change, most of them board seats. Proximity phrases keep the 8-Ks where the verb is about the chief executive.

```shell
curl -X POST https://www.quantisnow.com/api/v1/full-text \
  -H "Authorization: Bearer $QUANTISNOW_API_KEY" \
  -H "Content-Type: application/json" \
  -d '{
  "query": "items:\"5.02\" AND (\"retire chief executive officer\"~3 OR \"resign chief executive officer\"~3 OR \"step down chief executive officer\"~3 OR \"resignation chief executive officer\"~3)",
  "sort": "filedAt:desc",
  "size": 5,
  "facets": true
}'
```

## CFO departures in Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": 'items:"5.02" AND ("resign chief financial officer"~3 OR "depart chief financial officer"~3 '
                 'OR "step down chief financial officer"~3)',
        "sort": "filedAt:desc",
        "size": 25,
    },
)
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["companyName"])
    print("   ", f["highlights"][0] if f["highlights"] else "")
````

## In JavaScript

```javascript
const BASE = "https://www.quantisnow.com/api/v1";

async function search(endpoint, query, { size = 50, sort, from = 0 } = {}) {
  const res = await fetch(BASE + "/" + endpoint, {
    method: "POST",
    headers: {
      Authorization: "Bearer " + process.env.QUANTISNOW_API_KEY,
      "Content-Type": "application/json",
    },
    body: JSON.stringify({ query, size, from, sort }),
  });
  if (!res.ok) throw new Error((await res.json()).error.message);
  return res.json();
}

// Item 5.02 for a watchlist, last 30 days
const tickers = ["AAPL", "MSFT", "NVDA", "TSLA"];
const page = await search(
  "8-k/events",
  'items.item:"5.02" AND ticker:(' + tickers.join(" OR ") + ") AND filedAt:[2026-10-24 TO 2026-11-23]",
  { sort: "filedAt:desc" },
);
for (const f of page.data) console.log(f.ticker, f.filedAt, f.linkToFilingIndex);
```
