# How to screen 8-K filings by item number: auditor changes, delistings and bankruptcies

What each 8-K item number means, which ones traders watch, and how to query one item across every company with the 8-K API.

Read it with charts at https://www.quantisnow.com/sec-api/guides/8k-item-screener. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

A company files an 8-K within four business days of anything material, and every 8-K lists item numbers that say what kind of event it is. Searching by item is the fastest way to get one kind of event across the whole market.

## The items worth a screen

- 1.03 Bankruptcy or receivership: a Chapter 11 or 7 petition.
- 3.01 Notice of delisting: the exchange says the company breaks a listing rule, most often a share price under $1 for 30 days.
- 4.01 Change in certifying accountant: the auditor resigned or was dismissed. A resignation is the one to read.
- 4.02 Non-reliance on previous financial statements: a restatement is coming.
- 5.02 Departure or appointment of directors and officers: CEO and CFO changes.
- 1.05 Material cybersecurity incident: required since December 2023.

## Auditor changes this year

### Which companies changed auditors this year?

An auditor change, especially a resignation, often comes before a restatement.

```shell
curl -X POST https://www.quantisnow.com/api/v1/8-k/events \
  -H "Authorization: Bearer $QUANTISNOW_API_KEY" \
  -H "Content-Type: application/json" \
  -d '{
  "query": "items.item:\"4.01\" AND filedAt:[2026-01-01 TO 2026-11-24]",
  "from": 0,
  "size": 5,
  "sort": [{"filedAt": {"order": "desc"}}]
}'
```
