# Auditor resignations: search 8-K Item 4.01 filings

Find every 8-K Item 4.01 auditor change, then the ones where the auditor resigned or wouldn't stand for re-election, with the reasons in the filing's own words.

Read it with charts at https://www.quantisnow.com/sec-api/guides/auditor-resignations-8-k-4-01. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

An auditor change is filed under Item 4.01 within four business days. Most are routine: a company moves to a cheaper firm, or two firms merge. The ones to read are where the auditor quit, because an auditor rarely walks away from a paying client without a reason, and the 8-K has to say whether there were disagreements.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "items:\"4.01\" AND (resigned OR \"declined to stand\" OR \"decline to stand\")",
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## From the command line

```shell
npx quantisnow events auditor --days 365
npx quantisnow search 'items:"4.01" AND (resigned OR "declined to stand")' --forms 8-K --newest
```

## What to read in the filing

- Whether there were any disagreements with the auditor, and any reportable events such as a material weakness. Both have to be stated.
- The auditor's letter, attached as exhibit 16.1, saying whether it agrees with the company's account.
- How close the change is to a 10-K or 10-Q deadline. A resignation a few weeks before one often comes with an NT 10-K.
