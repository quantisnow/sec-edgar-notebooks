# Debt covenant breaches: who says they're not in compliance

Search 10-Qs and 8-Ks for companies that broke a loan covenant, whether lenders waived it, and what it cost.

Read it with charts at https://www.quantisnow.com/sec-api/guides/debt-covenant-breaches. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Loan agreements set minimums: liquidity, leverage, interest coverage. A company that misses one has to disclose it, and the lender can call the loan. Most breaches end in a waiver or an amendment, often with a fee or a higher rate; the filing says which.

## 10-Qs and 8-Ks that report a covenant breach

### Which companies say they were not in compliance with a debt covenant?

A covenant breach lets lenders demand repayment. Most get waived, but the waiver usually costs something, and some don't get one.

```shell
curl -X POST https://www.quantisnow.com/api/v1/full-text \
  -H "Authorization: Bearer $QUANTISNOW_API_KEY" \
  -H "Content-Type: application/json" \
  -d '{
  "query": "\"not in compliance\" AND covenant",
  "formTypes": ["10-Q", "10-K", "8-K"],
  "sort": "filedAt:desc",
  "size": 5
}'
```

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"not in compliance\" AND covenant",
        "formTypes": ["10-Q","10-K","8-K"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## From the command line

```shell
npx quantisnow search '"not in compliance" AND covenant' --forms 10-Q,8-K --newest
```

Add `AND waiver` to find the ones resolved, or `AND "event of default"` for the ones that weren't.
