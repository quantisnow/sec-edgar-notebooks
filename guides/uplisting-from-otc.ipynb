Moving from OTC to a national exchange opens a stock to funds and brokers that won't touch OTC names, and companies often raise money alongside it. They mention the plan in 10-Qs and S-1s, and the approval in an 8-K.

## Filings that mention uplisting

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "uplist OR uplisting OR uplisted",
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````
