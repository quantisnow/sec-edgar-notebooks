# SPAC deals and redemptions in 8-K filings

Find SPACs that signed a business combination and what their filings say about redemptions from the trust.

Read it with charts at https://www.quantisnow.com/sec-api/guides/spac-deals-redemptions. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

A SPAC holds its IPO money in a trust. When it signs a deal, public holders can redeem their shares for their slice of the trust. High redemptions leave the combined company with little of the cash the deal promised; the 8-Ks and 425 filings around the vote show where it's heading.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"business combination agreement\" AND redemptions",
        "formTypes": ["8-K"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````
