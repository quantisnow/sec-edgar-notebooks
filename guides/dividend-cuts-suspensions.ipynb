# Dividend cuts and suspensions in 8-K filings

Find companies that suspended or reduced their dividend, and what the 8-K says about why.

Read it with charts at https://www.quantisnow.com/sec-api/guides/dividend-cuts-suspensions. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

A dividend cut is a board admitting the cash is needed elsewhere, and income funds that own the stock for the payout may have to sell. The 8-K, or the release filed as its exhibit, usually gives the reason: debt, a downturn, or cash redirected to buybacks.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"suspension of quarterly cash dividend\"~3 OR \"suspend the quarterly dividend\"~3 OR \"reduce the quarterly dividend\"~3 OR \"suspension of its quarterly cash dividend\"~3",
        "formTypes": ["8-K"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## From the command line

```shell
npx quantisnow search '"suspension of quarterly cash dividend"~3 OR "reduce the quarterly dividend"~3' --forms 8-K --newest
```

Cuts are rare, so expect a short list. To spot them from the numbers instead, compare `cashFlow.dividendsPaid` across quarters in the financials endpoint.
