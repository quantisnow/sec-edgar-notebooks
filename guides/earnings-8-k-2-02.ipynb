## Guidance raises in Python

````python
import os, requests

res = requests.post("https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": 'items:"2.02" AND ("raises full-year guidance"~3 OR "raising guidance"~2 OR "increases outlook"~3) AND filedAt:>=2026-11-16',
        "sort": "filedAt:desc", "size": 50,
    }).json()
for f in res["data"]:
    print(f["ticker"], f["filedAt"][:16], f["linkToFilingIndex"])
````

## From the command line

```shell
npx quantisnow events earnings --days 7
npx quantisnow search 'items:"2.02" AND "lowers guidance"~3' --newest
```

The structured numbers (revenue, EPS) arrive with the 10-Q or 10-K a few weeks later, in the financials endpoint. The 8-K is where the words are first.
