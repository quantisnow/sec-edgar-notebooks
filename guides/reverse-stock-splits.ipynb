# Reverse stock splits: find them in 8-K filings

Every 8-K announcing a reverse split, with the ratio and effective date in the passage, plus how to tie it to a delisting notice.

Read it with charts at https://www.quantisnow.com/sec-api/guides/reverse-stock-splits. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

A reverse split cuts the share count to lift the price, nearly always to stay above an exchange's $1 minimum. The 8-K, usually under Item 5.03 for the charter amendment, gives the ratio and the date the stock starts trading split-adjusted.

## Reverse split 8-Ks, newest first

### Which companies filed an 8-K to put a reverse stock split into effect?

Item 5.03 is the charter amendment that makes the split happen, so this finds the effective splits with their ratios, not proxies asking for permission.

```shell
curl -X POST https://www.quantisnow.com/api/v1/full-text \
  -H "Authorization: Bearer $QUANTISNOW_API_KEY" \
  -H "Content-Type: application/json" \
  -d '{
  "query": "\"reverse stock split\" AND items:\"5.03\"",
  "sort": "filedAt:desc",
  "size": 5
}'
```

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"reverse stock split\" AND items:\"5.03\"",
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## From the command line

```shell
npx quantisnow search '"reverse stock split"' --forms 8-K --newest
```
