# CFO departures: 8-K Item 5.02 resignations

Find 8-Ks where the chief financial officer resigns or steps down, and read whether the company gives a reason.

Read it with charts at https://www.quantisnow.com/sec-api/guides/cfo-departures-8-k. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

A CFO leaving is filed under Item 5.02 within four business days. Companies often add that the departure "was not the result of any disagreement". When that sentence is missing, or the CFO leaves weeks before a 10-K is due, it's worth reading closely.

## CFO resignations in 8-K Item 5.02

### Which 8-Ks say the CFO is resigning or stepping down?

A CFO leaving without a stated reason, especially near a reporting deadline, is one of the more reliable warnings in the 8-K record.

```shell
curl -X POST https://www.quantisnow.com/api/v1/full-text \
  -H "Authorization: Bearer $QUANTISNOW_API_KEY" \
  -H "Content-Type: application/json" \
  -d '{
  "query": "items:\"5.02\" AND (\"resign chief financial officer\"~3 OR \"step down chief financial officer\"~3 OR \"depart chief financial officer\"~3 OR \"resignation chief financial officer\"~3)",
  "sort": "filedAt:desc",
  "size": 5
}'
```

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "items:\"5.02\" AND (\"resign chief financial officer\"~3 OR \"step down chief financial officer\"~3 OR \"depart chief financial officer\"~3 OR \"resignation chief financial officer\"~3)",
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````
