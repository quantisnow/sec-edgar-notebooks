# IPO risk factors: search S-1 and F-1 prospectuses

Search IPO prospectuses for the risk factors that matter, a history of losses, customer concentration, going-concern doubt, before a stock starts trading.

Read it with charts at https://www.quantisnow.com/sec-api/guides/ipo-s-1-risk-factors. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

A prospectus lists every risk the lawyers could think of, which makes the generic ones useless. The specific ones aren't: a history of losses, one customer, a going-concern paragraph, a controlling holder. Search for them across every S-1 and F-1 at once.

## Prospectuses that warn about a history of losses

### Which IPO prospectuses warn about a history of losses?

Risk factors in an S-1 or F-1 are written by lawyers, but the specific ones, a history of losses, one big customer, a going-concern note, say what the company is worried about.

```shell
curl -X POST https://www.quantisnow.com/api/v1/full-text \
  -H "Authorization: Bearer $QUANTISNOW_API_KEY" \
  -H "Content-Type: application/json" \
  -d '{
  "query": "\"history of losses\"~3 OR \"history of net losses\"~3",
  "formTypes": ["S-1", "S-1/A", "F-1", "F-1/A"],
  "sort": "filedAt:desc",
  "size": 5
}'
```

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"history of losses\"~3 OR \"history of net losses\"~3",
        "formTypes": ["S-1","S-1/A","F-1","F-1/A"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## From the command line

```shell
npx quantisnow search '"history of losses"~3' --forms S-1,S-1/A,F-1 --newest
npx quantisnow search '"substantial doubt" AND "going concern"' --forms S-1,F-1 --newest
```
