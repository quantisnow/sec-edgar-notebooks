# Bitcoin treasury companies: crypto on corporate balance sheets

Find companies that hold bitcoin or other digital assets as a treasury reserve, how they fund it, and the insider trades and fund positions in them.

Read it with charts at https://www.quantisnow.com/sec-api/themes/bitcoin-treasury-companies. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

A wave of small companies has turned into crypto holding companies: they raise money through ATMs, convertibles and PIPEs and buy bitcoin, ether or solana with it. Their 8-Ks report each purchase and each financing.

## A screener for this theme

Two calls: full-text search for the companies, then open-market insider buys in them. Change the query or the date range and run it again.

````python
import os, requests

API = "https://www.quantisnow.com/api/v1"
H = {"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"}

# 1. Companies whose filings talk about the theme
hits = requests.post(f"{API}/full-text", headers=H, json={
    "query": "\"bitcoin treasury\" OR \"digital asset treasury\" OR \"bitcoin reserve\" OR \"treasury reserve asset\"~3",
    "formTypes": ["10-K","10-Q","8-K"],
    "sort": "filedAt:desc", "size": 50,
}).json()["data"]
tickers = sorted({h["ticker"] for h in hits if h["ticker"]})
print(len(tickers), "companies:", ", ".join(tickers))

# 2. Open-market insider buys in those companies, last 90 days
if tickers:
    buys = requests.post(f"{API}/insider-trades", headers=H, json={
        "query": f"issuer.ticker:({' OR '.join(tickers[:40])}) AND transactionCodes:P AND filedAt:[2026-08-26 TO 2026-11-24]",
        "sort": "purchased.valueUsd:desc", "size": 50,
    }).json()["data"]
    for b in buys:
        print(b["issuer"]["ticker"], b["reportingOwner"]["name"], round(b["purchased"]["valueUsd"]))
````
