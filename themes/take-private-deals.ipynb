# Going-private and take-private deals in SEC filings

Find take-private deals through Schedule 13E-3 filings, merger proxies and 8-Ks, including management buyouts where insiders are on both sides.

Read it with charts at https://www.quantisnow.com/sec-api/themes/take-private-deals. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

When a controlling holder or management takes a company private, Rule 13e-3 requires a Schedule 13E-3 with extra disclosure about fairness, because insiders are on the buying side. Private equity take-privates show up in 8-Ks and merger proxies.

## A screener for this theme

Two calls: full-text search for the companies, then open-market insider buys in them. Change the query or the date range and run it again.

````python
import os, requests

API = "https://www.quantisnow.com/api/v1"
H = {"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"}

# 1. Companies whose filings talk about the theme
hits = requests.post(f"{API}/full-text", headers=H, json={
    "query": "\"going private\" OR \"Schedule 13E-3\" OR \"take-private\"",
    "formTypes": ["8-K","SC 13E3","PREM14A","DEFM14A","SC TO-T"],
    "sort": "filedAt:desc", "size": 50,
}).json()["data"]
tickers = sorted({h["ticker"] for h in hits if h["ticker"]})
print(len(tickers), "companies:", ", ".join(tickers))

# 2. Open-market insider buys in those companies, last 90 days
if tickers:
    buys = requests.post(f"{API}/insider-trades", headers=H, json={
        "query": f"issuer.ticker:({' OR '.join(tickers[:40])}) AND transactionCodes:P AND filedAt:[2026-08-26 TO 2026-11-24]",
        "sort": "purchased.valueUsd:desc", "size": 50,
    }).json()["data"]
    for b in buys:
        print(b["issuer"]["ticker"], b["reportingOwner"]["name"], round(b["purchased"]["valueUsd"]))
````
