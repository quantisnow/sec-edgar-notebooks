# Form RW: withdrawn registration statements and pulled IPOs

Form RW means a company pulled a registration: a postponed IPO, a canceled offering or a failed deal. Queries for recent withdrawals and their reasons.

Read it with charts at https://www.quantisnow.com/sec-api/forms/rw. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

A Form RW asks the SEC to withdraw a registration statement that hasn't been used. It's the paper trail of an IPO that didn't happen, a follow-on that was dropped, or an S-4 for a merger that fell apart.

## When it's filed

Whenever the company decides not to proceed; the SEC grants it unless it objects within 15 days.

## What traders read it for

- Pulled IPOs, which say a lot about the market for new issues.
- The stated reason, usually market conditions in one line.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"market conditions\" OR \"not to proceed\"~3 OR \"determined not to\"~3",
        "formTypes": ["RW"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).
