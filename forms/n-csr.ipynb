# N-CSR and N-CSRS: fund shareholder reports

Mutual fund and ETF annual and semiannual reports: holdings, performance and expenses. What's in N-CSR and queries for recent filings.

Read it with charts at https://www.quantisnow.com/sec-api/forms/n-csr. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Registered funds file their annual (N-CSR) and semiannual (N-CSRS) shareholder reports with the SEC, with certifications from the fund's officers. They carry the full schedule of investments, financial statements and the expense ratio.

## When it's filed

Within 10 days after the reports go to holders, which is within 60 days of the period end.

## What traders read it for

- A fund's full holdings at period end, including bonds and private positions that 13F leaves out.
- Fees and performance against the benchmark.

## N-CSR fund reports filings, newest first

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/filings",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "formType:(\"N-CSR\" OR \"N-CSRS\") AND filedAt:[2026-08-26 TO 2026-11-24]",
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
page = res.json()
print(page["total"]["value"], "matches")
for f in page["data"]:
    print(f["filedAt"][:10], f.get("ticker"), f["formType"], f.get("companyName"))
````


## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `marketCap`: Market cap in USD today, from the latest price. Micro cap is under 300000000, small cap under 2000000000. Example: marketCap:<300000000.
- `sector`: Sector: Basic Materials, Consumer Discretionary, Consumer Staples, Energy, Finance, Health Care, Industrials, Real Estate, Technology, Telecommunications or Utilities. Quote two-word names: sector:"Health Care".
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).
