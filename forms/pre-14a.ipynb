# PRE 14A, PREM14A and DEFM14A: preliminary and merger proxies

Why a company files a preliminary proxy, what a merger proxy reveals about a deal, and queries for recent PRE 14A and DEFM14A filings.

Read it with charts at https://www.quantisnow.com/sec-api/forms/pre-14a. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

A preliminary proxy (PRE 14A) is required, ten days ahead of the definitive one, when the vote covers more than routine items: a reverse split, an increase in authorized shares, a charter change. PREM14A and DEFM14A are the preliminary and definitive proxies for a merger vote.

## When it's filed

PRE 14A at least ten calendar days before the definitive proxy. A merger proxy follows the merger agreement by weeks and sets the meeting date.

## What traders read it for

- In a PRE 14A: the non-routine proposals, especially reverse split ranges and authorized share increases.
- In a merger proxy: the background of the merger, other bidders, the projections and the banker's valuation ranges.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"reverse stock split\" AND proposal",
        "formTypes": ["PRE 14A","DEF 14A"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).
