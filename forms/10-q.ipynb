# Form 10-Q: quarterly reports, and what to search in them

The 10-Q explained: deadlines, the sections traders read, and real queries for quarterly reports by date and by phrases like cash runway and covenant breaches.

Read it with charts at https://www.quantisnow.com/sec-api/forms/10-q. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

The 10-Q is the quarterly report for the first three fiscal quarters: unaudited financial statements, MD&A, updated risk factors and legal proceedings. The fourth quarter is covered by the 10-K.

## When it's filed

Within 40 days of quarter end for large accelerated and accelerated filers, 45 days for the rest. NT 10-Q buys five more days.

## What traders read it for

- Liquidity: the sentence saying how long current cash will fund operations.
- Subsequent events: financings, defaults and deals that happened after the quarter closed.
- Item 5, Other Information: officers and directors adopting or ending 10b5-1 trading plans.
- Changes in risk factors since the 10-K.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"sufficient to fund our operations\"~4 OR \"sufficient to fund our operating expenses\"~4",
        "formTypes": ["10-Q"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).

## Phrases worth searching inside it

- [sufficient to fund our operations~4](https://www.quantisnow.com/search?q=%22sufficient+to+fund+our+operations%22%7E4&kind=filing): `"sufficient to fund our operations"~4`
- [not in compliance AND covenant](https://www.quantisnow.com/search?q=%22not+in+compliance%22+AND+covenant&kind=filing): `"not in compliance" AND covenant`
- [subsequent events AND reverse stock split](https://www.quantisnow.com/search?q=%22subsequent+events%22+AND+%22reverse+stock+split%22&kind=filing): `"subsequent events" AND "reverse stock split"`
- [tariffs gross margin~12](https://www.quantisnow.com/search?q=%22tariffs+gross+margin%22%7E12&kind=filing): `"tariffs gross margin"~12`
