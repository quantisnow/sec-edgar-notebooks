# Form 6-K: reports from foreign private issuers

What foreign companies file on 6-K, why it matters for ADRs and foreign listings, and how to search 6-Ks for results, deals and meeting notices.

Read it with charts at https://www.quantisnow.com/sec-api/forms/6-k. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Foreign private issuers don't file 10-Qs or 8-Ks. Instead they furnish a 6-K with whatever they publish at home or to their exchange: quarterly or half-year results, acquisitions, management changes, shareholder meeting materials.

## When it's filed

Promptly after the company makes the information public in its home market or to its other exchange. There's no fixed calendar, so results arrive on the home market's schedule.

## What traders read it for

- Results from companies that only report half-yearly at home.
- Deal announcements and financings from Chinese, Israeli and European small caps listed in the US.
- Notices of reverse splits and Nasdaq deficiency letters, which foreign issuers report here instead of an 8-K.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"interim report\" OR \"half-year report\" OR \"unaudited interim\"~3",
        "formTypes": ["6-K"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## From the command line

```shell
npx quantisnow query filings 'formType:"6-K" AND ticker:TSM' --sort filedAt:desc
npx quantisnow search '"interim report"' --forms 6-K --newest
```

## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).
