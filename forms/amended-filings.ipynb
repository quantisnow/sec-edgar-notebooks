# Amended filings: 10-K/A, 10-Q/A, 8-K/A, 20-F/A and 6-K/A

Why companies amend reports, how to tell a routine amendment from a restatement, and queries for this week's amended filings.

Read it with charts at https://www.quantisnow.com/sec-api/forms/amended-filings. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

An amendment re-files part or all of a report with "/A" added to the form type. Most are routine: adding Part III pay disclosure to a 10-K, attaching financial statements of an acquired business to an 8-K, fixing an exhibit. Some restate numbers.

## When it's filed

Any time after the original. 8-K/As for acquisitions come within 71 days of the original 8-K.

## What traders read it for

- The explanatory note at the top, which says why it was filed.
- A 10-K/A or 10-Q/A that restates, which pairs with an Item 4.02 8-K.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"sole purpose of this amendment\"~3 OR \"purpose of this Amendment No.\"~3",
        "formTypes": ["10-K/A","10-Q/A","8-K/A","20-F/A"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).
