# DEF 14A: definitive proxy statements

What's in a proxy statement, when it's filed before the annual meeting, and queries for recent proxies, executive pay tables and shareholder proposals.

Read it with charts at https://www.quantisnow.com/sec-api/forms/def-14a. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

The definitive proxy statement goes to shareholders before a meeting. For an annual meeting it covers director elections, the auditor ratification, say-on-pay, executive compensation (the summary compensation table and pay versus performance), related-party transactions, and any shareholder proposals.

## When it's filed

At least 40 days before the meeting for notice-and-access delivery. For calendar-year companies that's March and April.

## What traders read it for

- Pay: what the CEO got and what it was tied to.
- Related-party transactions and director independence.
- Proposals that change the share count: an increase in authorized shares, a reverse split, a new equity plan.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"summary compensation table\"",
        "formTypes": ["DEF 14A"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `marketCap`: Market cap in USD today, from the latest price. Micro cap is under 300000000, small cap under 2000000000. Example: marketCap:<300000000.
- `sector`: Sector: Basic Materials, Consumer Discretionary, Consumer Staples, Energy, Finance, Health Care, Industrials, Real Estate, Technology, Telecommunications or Utilities. Quote two-word names: sector:"Health Care".
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).
