# Form S-3, S-3ASR and F-3: shelf registrations

How shelf registrations work, the difference between S-3, S-3ASR and F-3, and real queries for new shelves that often come before a stock sale.

Read it with charts at https://www.quantisnow.com/sec-api/forms/s-3. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

An S-3 is the short-form registration for companies that have been reporting for a year. Most are shelves: they register securities to sell later, in pieces, by prospectus supplement. Well-known seasoned issuers file S-3ASR, which is effective on filing. F-3 is the foreign version.

## When it's filed

Any time. A shelf lasts three years; companies often file a new one as the old one runs out, or right before they plan to raise.

## What traders read it for

- Who is selling: the company, or selling stockholders in a resale shelf.
- The dollar amount registered against the company's market value. Small companies are capped at a third of their public float a year (the baby-shelf rule).
- Whether an at-the-market sales agreement is filed alongside it.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"from time to time\" AND (shelf OR \"one or more offerings\")",
        "formTypes": ["S-3","S-3/A","S-3ASR"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````


## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `marketCap`: Market cap in USD today, from the latest price. Micro cap is under 300000000, small cap under 2000000000. Example: marketCap:<300000000.
- `sector`: Sector: Basic Materials, Consumer Discretionary, Consumer Staples, Energy, Finance, Health Care, Industrials, Real Estate, Technology, Telecommunications or Utilities. Quote two-word names: sector:"Health Care".
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).
