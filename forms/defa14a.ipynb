# DEFA14A and DFAN14A: additional proxy materials and proxy fights

Additional proxy materials from companies (DEFA14A) and from dissidents (DFAN14A), and how to spot a proxy fight from the filings.

Read it with charts at https://www.quantisnow.com/sec-api/forms/defa14a. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

DEFA14A is additional soliciting material from the company after its proxy: a letter to holders, a presentation, a response to a proxy adviser. DFAN14A is the same thing filed by anyone else, which almost always means an activist soliciting against management.

## When it's filed

As used, from the start of a solicitation to the meeting.

## What traders read it for

- DFAN14A filers: who is running a campaign, at which company.
- The back-and-forth in a contest, letter by letter.
- Merger votes, where companies use DEFA14A to answer opposition and adviser recommendations.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"universal proxy card\"",
        "formTypes": ["DEFA14A","DFAN14A","DEF 14A","PRE 14A"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````


## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).
