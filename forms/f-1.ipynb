# Form F-1: IPO registrations by foreign companies

The foreign issuer's S-1: what an F-1 discloses, why so many small Asian IPOs use it, and queries for recent F-1s and their China and VIE language.

Read it with charts at https://www.quantisnow.com/sec-api/forms/f-1. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

An F-1 is the registration statement a foreign private issuer uses for an IPO or offering in the US. It has the same core as an S-1, with foreign-issuer items: home-country regulation, currency controls, and for many filers, a holding company structure in the Cayman Islands or BVI.

## When it's filed

Like an S-1, at the issuer's chosen time, with F-1/A amendments until it's declared effective.

## What traders read it for

- Where the operating business is and how the listed company controls it, contract (VIE) or equity.
- The size of the offering against the company's revenue. Many recent small-cap F-1s raise a few million dollars.
- Lock-up terms and who holds the shares after the IPO.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"People's Republic of China\" AND (\"Holding Foreign Companies Accountable Act\" OR HFCAA)",
        "formTypes": ["10-K","10-Q","20-F","F-1","F-1/A"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `marketCap`: Market cap in USD today, from the latest price. Micro cap is under 300000000, small cap under 2000000000. Example: marketCap:<300000000.
- `sector`: Sector: Basic Materials, Consumer Discretionary, Consumer Staples, Energy, Finance, Health Care, Industrials, Real Estate, Technology, Telecommunications or Utilities. Quote two-word names: sector:"Health Care".
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).
