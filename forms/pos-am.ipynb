# POS AM: post-effective amendments to registration statements

What a post-effective amendment does, why most deregister unsold securities, and queries for recent POS AM filings.

Read it with charts at https://www.quantisnow.com/sec-api/forms/pos-am. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

A POS AM changes a registration statement that's already effective: to update the prospectus with a new 10-K, to add or remove securities, or to deregister what wasn't sold. After a merger, the acquired company files a batch of them to close out its old shelves.

## When it's filed

As needed; a shelf's prospectus must be updated at least yearly, and deregistration follows a merger closing.

## What traders read it for

- A cluster of POS AMs from one company, which usually means a merger just closed.
- Resale shelves being kept alive for selling holders.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/filings",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "formType:(\"POS AM\") AND filedAt:[2026-10-25 TO 2026-11-24]",
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
page = res.json()
print(page["total"]["value"], "matches")
for f in page["data"]:
    print(f["filedAt"][:10], f.get("ticker"), f["formType"], f.get("companyName"))
````

## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).
