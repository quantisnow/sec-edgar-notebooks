# Form 20-F and 40-F: annual reports from foreign issuers

The foreign issuer's annual report: what's in a 20-F, when the 40-F applies, and queries for recent filings and China and VIE disclosures.

Read it with charts at https://www.quantisnow.com/sec-api/forms/20-f. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

A 20-F is a foreign private issuer's annual report: the business, risk factors, operating review, financial statements (under IFRS or US GAAP), major shareholders and related-party dealings. Canadian companies eligible for the multijurisdictional system file a 40-F instead, wrapping their Canadian annual filings.

## When it's filed

Within four months of the fiscal year end. Most foreign issuers have December year ends, so April is the busy month.

## What traders read it for

- Item 7, major shareholders: who controls the company, often a founder or a state entity.
- For China-based companies: the VIE structure, cash transfer restrictions and the HFCAA discussion.
- Item 16F, change in certifying accountant, the foreign version of an 8-K Item 4.01.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"variable interest entity\" AND (PRC OR \"People's Republic of China\")",
        "formTypes": ["10-K","10-Q","20-F","F-1","F-1/A"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## From the command line

```shell
npx quantisnow query filings 'formType:("20-F" OR "40-F")' --sort filedAt:desc -n 20
npx quantisnow search '"variable interest entity"' --forms 20-F --newest
```

## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).
