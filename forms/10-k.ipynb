# Form 10-K: the annual report, and how to search it

What a 10-K contains, when it's due, what traders read first, and real queries for 10-K filings by date, company and the words inside them.

Read it with charts at https://www.quantisnow.com/sec-api/forms/10-k. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

The 10-K is a company's annual report to the SEC: the business, risk factors, management's discussion of results (MD&A), audited financial statements, and the auditor's opinion. Amendments are filed as 10-K/A, often to add the executive pay section (Part III) after the proxy.

## When it's filed

Within 60 days of the fiscal year end for large accelerated filers, 75 days for accelerated filers and 90 days for everyone else. A company that can't make it files an NT 10-K for 15 more days.

## What traders read it for

- Risk factors that are new since last year. Lawyers add them when something has changed.
- MD&A on liquidity: how long the cash lasts and what the debt covenants require.
- The auditor's opinion: a going-concern paragraph, a critical audit matter, or an adverse opinion on internal controls.
- Customer concentration and segment detail that the earnings release leaves out.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"critical audit matter\"",
        "formTypes": ["10-K"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `marketCap`: Market cap in USD today, from the latest price. Micro cap is under 300000000, small cap under 2000000000. Example: marketCap:<300000000.
- `sector`: Sector: Basic Materials, Consumer Discretionary, Consumer Staples, Energy, Finance, Health Care, Industrials, Real Estate, Technology, Telecommunications or Utilities. Quote two-word names: sector:"Health Care".
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).

## Phrases worth searching inside it
