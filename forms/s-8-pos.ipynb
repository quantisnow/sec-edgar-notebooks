# S-8 POS: post-effective amendments to employee plan registrations

Most S-8 POS filings deregister plan shares after an acquisition closes. How to use them as a merger-closing signal, with recent filings.

Read it with charts at https://www.quantisnow.com/sec-api/forms/s-8-pos. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

An S-8 POS amends an S-8. Nearly all of them deregister the unissued shares under the acquired company's equity plans when a merger closes, so they arrive in batches on closing day.

## When it's filed

On or right after the closing of an acquisition, or when a plan is terminated.

## What traders read it for

- Merger closings, filed under the target's name, often before the press release.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/filings",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "formType:(\"S-8 POS\") AND filedAt:[2026-10-25 TO 2026-11-24]",
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
page = res.json()
print(page["total"]["value"], "matches")
for f in page["data"]:
    print(f["filedAt"][:10], f.get("ticker"), f["formType"], f.get("companyName"))
````

## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).
