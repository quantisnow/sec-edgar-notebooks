# Form S-1: IPO and resale registration statements

What an S-1 registers, how amendments track an IPO toward pricing, and real queries for new S-1 filings and the risk factors and use of proceeds inside them.

Read it with charts at https://www.quantisnow.com/sec-api/forms/s-1. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

An S-1 registers securities for sale by a US company that can't use the short-form S-3: an IPO, a follow-on by a small company, or a resale registration for shares that private investors bought. Each S-1/A amendment responds to SEC comments or fills in the price range.

## When it's filed

Whenever the company is ready. An IPO S-1 is often filed confidentially first and made public at least 15 days before the roadshow; amendments follow until the SEC declares it effective.

## What traders read it for

- Whether it's an IPO or a resale. A resale S-1 registers shares for selling holders, often after a PIPE or convertible note, and the company gets no money.
- Use of proceeds, dilution per share and the capitalization table.
- Risk factors specific to the company, not the boilerplate: a history of losses, one big customer, a going-concern note.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"we intend to use the net proceeds\"~4",
        "formTypes": ["S-1","S-1/A","F-1","F-1/A"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `marketCap`: Market cap in USD today, from the latest price. Micro cap is under 300000000, small cap under 2000000000. Example: marketCap:<300000000.
- `sector`: Sector: Basic Materials, Consumer Discretionary, Consumer Staples, Energy, Finance, Health Care, Industrials, Real Estate, Technology, Telecommunications or Utilities. Quote two-word names: sector:"Health Care".
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).
