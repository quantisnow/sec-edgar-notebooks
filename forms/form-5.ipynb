# Form 5: annual statements of insider transactions

What lands on a Form 5 instead of a Form 4, the 45-day deadline after year end, and queries for recent Form 5 filings.

Read it with charts at https://www.quantisnow.com/sec-api/forms/form-5. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Form 5 is the annual catch-up for insiders: small acquisitions and gifts that were allowed to wait, and any transactions that should have been on a Form 4 but weren't. Many insiders never file one because they have nothing to report.

## When it's filed

Within 45 days after the company's fiscal year end.

## What traders read it for

- Gifts of stock, which have been the subject of SEC enforcement when timed before bad news.
- Late-reported trades, a sign of weak compliance.

## Form 5 filings, newest first

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/insider-trades",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "formType:5 AND filedAt:[2026-08-26 TO 2026-11-24]",
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
page = res.json()
print(page["total"]["value"], "matches")
for f in page["data"]:
    print(f["filedAt"][:10], f.get("ticker"), f["formType"], f.get("companyName"))
````

## Fields the API returns

- `accessionNo`: EDGAR accession number.
- `formType`: 3, 4, 5 or 144, with /A on amendments (4/A).
- `amendment`: True for 4/A and the like.
- `restatesEarlier`: True when this amendment repeats trades from a filing already in the results.
- `filedAt`: Acceptance time, US Eastern.
- `transactionDate`: Trade date reported on the form (planned sale date on a 144).
- `issuer.cik`: Company CIK.
- `issuer.name`: Company name.
- `issuer.ticker`: Company ticker.
- `reportingOwner.cik`: The insider's own CIK. Stable across companies and name spellings.
- Full list on the [Insider trading page](https://www.quantisnow.com/sec-api/insider-trading).
