# Form 25-NSE: exchange notices of delisting

What a 25-NSE means, how it differs from an 8-K delisting notice, and queries for securities struck from Nasdaq and NYSE.

Read it with charts at https://www.quantisnow.com/sec-api/forms/25-nse. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

A 25-NSE is the exchange's own filing that it will strike a class of securities from listing. Most follow completed mergers, redeemed preferreds and notes, or SPAC liquidations; some follow a company that failed listing rules and exhausted its appeals.

## When it's filed

Ten days before the delisting takes effect.

## What traders read it for

- The end point of a merger: the target's stock stops trading.
- Involuntary delistings, which move a stock to the OTC markets.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/filings",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "formType:(\"25-NSE\") AND filedAt:[2026-10-25 TO 2026-11-24]",
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
page = res.json()
print(page["total"]["value"], "matches")
for f in page["data"]:
    print(f["filedAt"][:10], f.get("ticker"), f["formType"], f.get("companyName"))
````

## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).
