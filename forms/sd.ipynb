# Form SD: conflict minerals and resource payments

What Form SD reports, which companies file it, and queries for recent conflict minerals and resource extraction disclosures.

Read it with charts at https://www.quantisnow.com/sec-api/forms/sd. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Form SD carries specialized disclosures that don't fit elsewhere: the conflict minerals report (tin, tantalum, tungsten and gold from the Congo region in a company's products) and, for oil, gas and mining companies, payments made to governments.

## When it's filed

By May 31 each year for the prior calendar year.

## What traders read it for

- Supply chain exposure for electronics and industrial makers.
- Government payments by country for resource companies.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/filings",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "formType:(\"SD\") AND filedAt:[2026-05-28 TO 2026-11-24]",
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
page = res.json()
print(page["total"]["value"], "matches")
for f in page["data"]:
    print(f["filedAt"][:10], f.get("ticker"), f["formType"], f.get("companyName"))
````

## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).
