# PREC14A and DEFC14A: contested proxy statements

Proxy statements filed in a contested election, by the company or the dissident. How to follow a proxy fight from the filings, with recent examples.

Read it with charts at https://www.quantisnow.com/sec-api/forms/contested-proxies. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

The C forms are proxy statements in a contested solicitation: a dissident running its own director slate, or opposing a deal. Either side can file them, and each side's nominees appear on the universal proxy card.

## When it's filed

Preliminary (PREC14A) at least ten days before the definitive (DEFC14A), with materials filed through to the meeting.

## What traders read it for

- Each side's case for its nominees.
- Whether the contest settled: a cooperation agreement in an 8-K usually ends it.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"universal proxy card\"",
        "formTypes": ["DEFA14A","DFAN14A","DEF 14A","PRE 14A"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````


## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).
