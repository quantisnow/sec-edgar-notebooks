# Form 8-K: current reports and how to filter them

What triggers an 8-K, the four-business-day deadline, and real queries by item number, company and text, with a link to every item's own guide.

Read it with charts at https://www.quantisnow.com/sec-api/forms/8-k. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

An 8-K reports a material event between periodic reports. Each one lists the items it covers, from Item 1.01 (a material agreement) to 9.01 (exhibits), and the press release or contract is usually attached as an exhibit. Corrections are filed as 8-K/A.

## When it's filed

Within four business days of the event. Item 7.01 (Regulation FD) and 2.02 (results) material is often furnished the same day as the press release.

## What traders read it for

- The item numbers: they say what happened before you read a word. See the [item index](https://www.quantisnow.com/sec-api/forms/8-k-items).
- Exhibit 99.1, the release, and exhibit 10.x, the actual contract.
- Timing: an 8-K filed late on a Friday is a long-standing way to bury news.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"entry into a material definitive agreement\"",
        "formTypes": ["8-K"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## Fields the API returns

- `accessionNo`: EDGAR accession number.
- `formType`: 8-K or 8-K/A.
- `filedAt`: Acceptance time, US Eastern.
- `cik`: Filer CIK.
- `companyName`: Company name.
- `ticker`: Ticker.
- `items`: Each item reported: `item` ("5.02") and its `title`.
- Full list on the [8-K events page](https://www.quantisnow.com/sec-api/8-k-events).
