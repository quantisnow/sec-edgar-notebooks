# Form 3: initial statements of insider ownership

When a new director, officer or 10% holder files a Form 3, what it shows, and queries for this week's new insiders.

Read it with charts at https://www.quantisnow.com/sec-api/forms/form-3. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Form 3 is the first filing a person makes on becoming an insider: a director, an officer, or a holder of more than 10%. It lists what they own on that day, shares and derivatives, and starts the clock for Form 4s.

## When it's filed

Within ten days of becoming an insider, or by the effective date of an IPO registration.

## What traders read it for

- New board members and officers, often before the company announces them prominently.
- New 10% holders, which can point to a financing or a stake building.
- Starting holdings, to judge the size of later purchases and sales.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/insider-trades",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "formType:3 AND filedAt:[2026-11-17 TO 2026-11-24]",
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
page = res.json()
print(page["total"]["value"], "matches")
for f in page["data"]:
    print(f["filedAt"][:10], f.get("ticker"), f["formType"], f.get("companyName"))
````

## Fields the API returns

- `accessionNo`: EDGAR accession number.
- `formType`: 3, 4, 5 or 144, with /A on amendments (4/A).
- `amendment`: True for 4/A and the like.
- `restatesEarlier`: True when this amendment repeats trades from a filing already in the results.
- `filedAt`: Acceptance time, US Eastern.
- `transactionDate`: Trade date reported on the form (planned sale date on a 144).
- `issuer.cik`: Company CIK.
- `issuer.name`: Company name.
- `issuer.ticker`: Company ticker.
- `reportingOwner.cik`: The insider's own CIK. Stable across companies and name spellings.
- Full list on the [Insider trading page](https://www.quantisnow.com/sec-api/insider-trading).
