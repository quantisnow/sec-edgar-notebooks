# Schedule 13D: activist and controlling stakes over 5%

Who files a 13D, the new five-business-day deadline, what Item 4 reveals about intentions, and queries for new 13Ds and amendments.

Read it with charts at https://www.quantisnow.com/sec-api/forms/schedule-13d. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Anyone who acquires more than 5% of a company's voting stock with an intent to influence or control it files a Schedule 13D. Item 4, purpose of the transaction, says what they plan; Item 5 gives the stake and recent trades. Amendments (13D/A) report material changes, including any 1% move.

## When it's filed

Within five business days of crossing 5% (it was ten until 2024), and amendments within two business days of a material change.

## What traders read it for

- Item 4: a letter to the board, a nomination, a plan to push a sale.
- Item 5(c): every trade in the last 60 days, with prices.
- Exhibits: the letters themselves, and any cooperation or standstill agreement.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"intend to nominate\"~3 OR \"intends to nominate\"~3 OR \"delivered a letter\"~4 OR \"nomination notice\"~3 OR \"proxy contest\"",
        "formTypes": ["SCHEDULE 13D","SCHEDULE 13D/A"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## From the command line

```shell
npx quantisnow query ownership 'formType:13D' --sort filedAt:desc -n 20
npx quantisnow search '"delivered a letter"' --forms 'SCHEDULE 13D,SCHEDULE 13D/A' --newest
```

## Fields the API returns

- `accessionNo`: EDGAR accession number.
- `formType`: 13D or 13G. Amendments carry the same type.
- `filedAt`: Acceptance time, US Eastern.
- `filer.cik`: CIK of the person or fund holding the stake.
- `filer.name`: Filer name.
- `subject.cik`: CIK of the company the stake is in.
- `subject.name`: Company name.
- `subject.ticker`: Company ticker.
- Full list on the [13D and 13G stakes page](https://www.quantisnow.com/sec-api/beneficial-ownership).
