# Form 8-A12B: registering securities on an exchange

What an 8-A12B means, why it comes right before a listing, uplisting or spin-off, and queries for recent exchange registrations.

Read it with charts at https://www.quantisnow.com/sec-api/forms/8-a12b. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Form 8-A registers a class of securities under the Exchange Act. The 12(b) version is for a national exchange such as Nasdaq or NYSE, the 12(g) version for securities not on an exchange. It's short; the point is what it signals.

## When it's filed

Shortly before trading starts: the day of or before an IPO's pricing, before an uplisting from the OTC, or before a new class of preferreds or notes lists.

## What traders read it for

- Upcoming listings from companies moving up from the OTC markets.
- New exchange-listed notes and preferreds from banks, REITs and utilities.
- Spin-offs, where the new company registers its stock.

## 8-A12B listings filings, newest first

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/filings",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "formType:(\"8-A12B\" OR \"8-A12G\") AND filedAt:[2026-10-25 TO 2026-11-24]",
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
page = res.json()
print(page["total"]["value"], "matches")
for f in page["data"]:
    print(f["filedAt"][:10], f.get("ticker"), f["formType"], f.get("companyName"))
````

## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).
