# Form S-4: merger and exchange-offer registrations

When a stock deal needs an S-4, what it reveals about the merger, and queries for recent S-4s and their exchange ratios.

Read it with charts at https://www.quantisnow.com/sec-api/forms/s-4. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

An S-4 registers shares a company will issue in a merger or exchange offer. It's usually also the proxy statement for the target's vote, so it carries the full deal: the background of the merger, the fairness opinions with the bankers' valuation ranges, projections and risk factors. F-4 is the foreign version.

## When it's filed

Weeks after the merger agreement is signed. The vote follows once the SEC declares it effective, after S-4/A amendments.

## What traders read it for

- The background section: other bidders, earlier prices offered, how the price moved.
- The banker's analysis and the management projections behind it.
- Termination fees and the conditions to closing.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"exchange ratio\"",
        "formTypes": ["S-4","S-4/A","425"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## From the command line


## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `marketCap`: Market cap in USD today, from the latest price. Micro cap is under 300000000, small cap under 2000000000. Example: marketCap:<300000000.
- `sector`: Sector: Basic Materials, Consumer Discretionary, Consumer Staples, Energy, Finance, Health Care, Industrials, Real Estate, Technology, Telecommunications or Utilities. Quote two-word names: sector:"Health Care".
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).
