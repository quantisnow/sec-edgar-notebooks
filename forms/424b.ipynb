# 424B3, 424B4 and 424B5: prospectuses that price and sell securities

What each 424B rule means, how 424B4 marks an IPO pricing and 424B5 a shelf takedown, and queries for recent prospectuses with the offering price.

Read it with charts at https://www.quantisnow.com/sec-api/forms/424b. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

Prospectuses filed under Rule 424(b) are where registered offerings actually happen. A 424B4 is the final prospectus for an IPO or follow-on, with the price. A 424B5 is a supplement off a shelf: an ATM program, an overnight offering, a registered direct. A 424B3 updates a prospectus with new facts, common for resale shelves, merger prospectuses and non-traded REITs. A 424B7 adds selling holders to a resale shelf. (424B2, mostly bank structured notes, isn't in our data.)

## When it's filed

Within two business days after pricing or first use.

## What traders read it for

- The price, the share count and the discount against the last close.
- For ATMs, the size of the program against market value.
- Warrants sold alongside shares, which mean more dilution later.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"public offering price\"~3 AND underwriters",
        "formTypes": ["424B4"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````

## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).
