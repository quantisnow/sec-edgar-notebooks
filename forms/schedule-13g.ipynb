# Schedule 13G: passive stakes over 5%

Who can file the short-form 13G instead of a 13D, its deadlines, and queries for new passive stakes by filer and company.

Read it with charts at https://www.quantisnow.com/sec-api/forms/schedule-13g. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

A Schedule 13G is the short form for holders above 5% who don't intend to influence the company: index funds, other institutions buying in the normal course, and passive investors under 20%. It gives the stake and little else.

## When it's filed

Since 2024: institutions within 45 days of the quarter end in which they crossed 5%, other passive holders within five business days. Amendments follow quarterly for material changes.

## What traders read it for

- Unfamiliar names in small companies: a new 13G from a hedge fund is a stake worth knowing about.
- A switch from 13G to 13D, which means the holder has decided to be active.
- Exits: an amendment showing the stake fell below 5%.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/ownership",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "formType:13G AND filedAt:[2026-11-17 TO 2026-11-24]",
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
page = res.json()
print(page["total"]["value"], "matches")
for f in page["data"]:
    print(f["filedAt"][:10], f.get("ticker"), f["formType"], f.get("companyName"))
````

## From the command line

```shell
npx quantisnow query ownership 'formType:13G AND subject.ticker:GME' --sort filedAt:desc
```

## Fields the API returns

- `accessionNo`: EDGAR accession number.
- `formType`: 13D or 13G. Amendments carry the same type.
- `filedAt`: Acceptance time, US Eastern.
- `filer.cik`: CIK of the person or fund holding the stake.
- `filer.name`: Filer name.
- `subject.cik`: CIK of the company the stake is in.
- `subject.name`: Company name.
- `subject.ticker`: Company ticker.
- Full list on the [13D and 13G stakes page](https://www.quantisnow.com/sec-api/beneficial-ownership).
