# Tender offers: SC TO-T, SC TO-I, SC TO-C and SC 14D9

Third-party and issuer tender offers and the target's response: which form is which, what to read, and queries for recent tender offer filings.

Read it with charts at https://www.quantisnow.com/sec-api/forms/tender-offers. More notebooks: https://github.com/quantisnow/sec-edgar-notebooks. Set `QUANTISNOW_API_KEY` before running; make a key at https://www.quantisnow.com/sec-api/keys.

A tender offer is a public offer to buy shares directly from holders. SC TO-T is filed by a third party, SC TO-I by a company buying back its own shares (including Dutch auctions), SC TO-C for communications before the offer starts. The target answers with a Schedule 14D-9 recommending holders accept, reject or stay neutral.

## When it's filed

On the day the offer starts. It must stay open at least 20 business days; the target's 14D-9 is due within ten business days.

## What traders read it for

- The price, the minimum tender condition and the expiration date.
- In a 14D-9: the board's recommendation and the reasons, plus the background of talks.
- In an issuer tender: the price range of a Dutch auction and the number of shares sought.

## In Python

````python
import os, requests

res = requests.post(
    "https://www.quantisnow.com/api/v1/full-text",
    headers={"Authorization": f"Bearer {os.environ['QUANTISNOW_API_KEY']}"},
    json={
        "query": "\"offer to purchase\" AND expire",
        "formTypes": ["SC TO-T","SC TO-I","SC TO-C","SC 14D9"],
        "sort": "filedAt:desc",
        "size": 50,
    },
    timeout=30,
)
res.raise_for_status()
for f in res.json()["data"]:
    print(f["filedAt"][:10], f["ticker"], f["formType"], f["companyName"])
    if f["highlights"]:
        print("   ", f["highlights"][0].replace("<em>", "").replace("</em>", ""))
````


## Fields the API returns

- `accessionNo`: EDGAR accession number, unique per filing, e.g. 0001045810-26-000101.
- `formType`: Form as EDGAR names it: 10-K, 8-K, S-1, 4, SCHEDULE 13D/A, D, NT 10-Q.
- `filedAt`: Acceptance time in US Eastern time, ISO 8601 with offset.
- `cik`: Filer CIK, zero-padded to 10 digits.
- `companyName`: Filer name as it appears on the filing.
- `ticker`: Primary ticker of the issuer, when it trades.
- `items`: 8-K item numbers, e.g. ["2.02", "9.01"]. Empty for other forms.
- `description`: One-line description of the filing.
- Full list on the [Filings search page](https://www.quantisnow.com/sec-api/filings-search).
